A GGUF file is a binary format designed for storing large language models (LLMs). [ref](https://huggingface.co/docs/hub/gguf)

Model characteristics:
- Mistral-7B is the underlying ~7B-parameter general-purpose language model. [ref](https://mistral.ai/news/announcing-mistral-7b/)
    - Model parameters are the fundamental building blocks that determine LLM capabilities, performance, and resource requirements. Designations like "7B," "15B," or "70B" in model names refer to billions of parameters, the trainable weights and connections that enable the model to understand and generate text. [ref](https://local-ai-zone.github.io/guides/what-is-ai-model-3b-7b-30b-parameters-guide-2025.html)
- Instruct-v0.2 means it was instruction-fine-tuned to respond to questions, instructions, dialogue, summarization, extraction, etc.
- Mistral Instruct expects instructions in a specific form that has a "begin of sentence ID" : [ref](https://huggingface.co/mistralai/Mistral-7B-Instruct-v0.2/blob/refs%2Fpr%2F128/README.md)
    ```text
    <s>[INST] Some instruction [/INST]
    ```

Quantized model:
- GGUF Q6_K is not a different trained model. It's a quantized representation designed for efficient inference through llama.cpp.
- Quantization is the process of reducing the numerical precision of a model's weights from their training precision to a lower-bit representation, incredibly reducing the model's memory footprint. [ref](https://canitrun.dev/guides/quantization-explained/)
- This Q6_K file is about 5.94 GB and is characterized as having extremely low quality loss compared with the unquantized weights.

````python
# Download the model file from its Hugging Face Hub repository
model_path = hf_hub_download(
    repo_id=model_name_or_path,
    filename=model_basename
)
````

The type of model itself is the largest influence on response quality.

For example, going from an older general 7B to a better-trained model can dwarf changes to top_p (cumulative probability thershold) or top_k (fixed number of candidate tokens).

````python
# # Create an instance of the selected LLama model on GPU
# llm = Llama(
#     model_path=model_path,
#     n_ctx=default_n_ctx,
#     n_gpu_layers=default_n_gpu_layers,
#     n_batch=default_n_batch,
#     verbose=False
# )

# Create an instance of the selected LLama model on CPU
llm = Llama(
    model_path=model_path,
    n_ctx=default_n_ctx,
    n_gpu_layers=default_n_gpu_layers,
    n_batch=default_n_batch,
    n_threads=os.cpu_count(),
    n_threads_batch=os.cpu_count(),
    verbose=False
)
````

n_ctx:
- This defines the maximum working context window, used by a KV memory cache.
- The values set here (2300 for GPU; 1024 for CPU) are actually quite restrictive for RAG, but are in consideration of the low fidelity runtime used by freemium Google Colab.
- More context costs memory and can increase processing time.
- Mistral v0.2 itself supports a much larger context window.
- In general, though, for PDF/RAG work, something like `n_ctx=8192` is a much more sensible starting point.

n_gpu_layers:
- This controls how much of the model gets put on the GPU.
- Current llama-cpp-python supports `n_gpu_layers=-1` to offload all available model layers to the GPU, but only when installed with `CMAKE_ARGS="-DGGML_CUDA=on"`.
- This more layers offloaded to the GPU, the faster the model's performance.
- Mistral 7B only has 32 transformer layers, so in practice, with this set to 32, it's already effectively trying to offload essentially the entire model, so let's just use -1 on GPU.

n_batch:
- This primarily affects prompt ingestion speed.
- Larger values may improve prompt processing throughput while consuming more memory.
- Doesn't affect intelligence.

LLMs have developed a myriad of generation parameters, using at the very least max_tokens, temperature, and top_p. [ref](https://amitray.com/llm-parameters-temperature-top-p-top-k-guide/)

max_tokens:
- Controls maximum output length.
- Doesn't affect intelligence.

temperature:
- Affects token randomness.
- For factual medical prompts, deterministic/greedy generation (`temperature=0`) is desirable because you're trying to measure prompt differences instead of randomness.

top_p and top_k:
- These are sampling controls.
- Since we want more deterministic/greedy responses, manipulating these isn't particularly useful for this project.

````python
# Generate a response specifically from an instruction-tuned/chat model
def generate_response(
    prompt,
    max_tokens=default_max_tokens,
    temperature=0,
    top_p=0.95,
    top_k=40
):
    if max_tokens is None:
        max_tokens = default_max_tokens

    generation_args = {
        "max_tokens": max_tokens,
        "temperature": temperature
    }

    # Sampling controls are only relevant when sampling is enabled
    if temperature > 0:
        generation_args.update({
            "top_p": top_p,
            "top_k": top_k
        })

    output = llm.create_chat_completion(
        messages=[
            {"role": "user", "content": prompt}
        ],
        **generation_args
    )

    return output["choices"][0]["message"]["content"].strip()
````

### Embedding

````python
# Define the embedding model used to embed our vector store
# Use cached one if it exists
# NOTE: Cached model must be the same as the one defined here
embedding_model = SentenceTransformerEmbeddings(
    model_name="NeuML/biomedbert-small-embeddings",
    cache_folder=embedding_model_dir,
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)
````

Choosing the right embedding model is important, as they each have a certain "max sequence" size, which correlates with how big our chunk size is.

biomedbert-small-embeddings model:
- This is a 22.7M-parameter, 384-dimensional Sentence Transformers model.
- It has a 512-token maximum sequence length.
- Specifically trained/distilled for biomedical semantic retrieval.
- Quite perfect for this project. [ref](https://huggingface.co/NeuML/biomedbert-small-embeddings)

````python
# Check embedding semantics if no existing vector database
if build_vectorstore:
    embedding_1 = embedding_model.embed_query(document_chunks[0].page_content)
    embedding_2 = embedding_model.embed_query(document_chunks[1].page_content)
    print("Dimension of the embedding vector: ", len(embedding_1))
    print(len(embedding_1)==len(embedding_2))
````

````python
if build_vectorstore:
    vectorstore = Chroma.from_documents(
        documents=document_chunks,
        embedding=embedding_model,
        persist_directory=db_dir
    )

    print(
        f"Vectorstore created: "
        f"{vectorstore._collection.count():,} chunks"
    )

    # Create local ZIP
    local_zip_path = shutil.make_archive(
        "/content/medical_db",
        "zip",
        root_dir=db_dir
    )

    # Copy it to persistent Google Drive storage
    shutil.copy2(
        local_zip_path,
        db_zip_path
    )

    print(f"Saved vectorstore to:\n{db_zip_path}")
````

````python
# Generate a RAG-grounded response using retrieved document context
def generate_rag_response(
    user_input,
    k=3,
    max_tokens=None,
    temperature=0,
    top_p=0.95,
    top_k=40
):
    if max_tokens is None:
        max_tokens = default_max_tokens

    # Retrieve the most relevant document chunks.
    # Create a temporary retriever so k can vary per request.
    query_retriever = vectorstore.as_retriever(
        search_kwargs={"k": k}
    )
    relevant_document_chunks = query_retriever.invoke(user_input)

    # Preserve clear boundaries between retrieved chunks.
    context_for_query = "\n\n---\n\n".join(
        document.page_content.strip()
        for document in relevant_document_chunks
    )

    # Insert the retrieved context and question into the user prompt template.
    user_message = qna_user_message_template.format(
        context=context_for_query,
        question=user_input
    )

    generation_args = {
        "max_tokens": max_tokens,
        "temperature": temperature
    }

    # Sampling controls are only useful when sampling is enabled.
    if temperature > 0:
        generation_args.update({
            "top_p": top_p,
            "top_k": top_k
        })

    try:
        output = llm.create_chat_completion(
            messages=[
                {
                    "role": "system",
                    "content": qna_system_message
                },
                {
                    "role": "user",
                    "content": user_message
                }
            ],
            **generation_args
        )

        return output["choices"][0]["message"]["content"].strip()

    except Exception as e:
        return f"Error generating response: {e}"
````

- High (0.9–1.5) → creative, diverse

## Output Evaluation

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answeres generated to the question from the previous section.

- We are using the same Mistral model for evaluation, so basically here the llm is rating itself on how well he has performed in the task.
- On CPU, I am using the fastest model, which is SmolM2 1.7B.

Groundedness should evaluate generation: is the answer actually supported by the retrieved context?

Retrieval relevance should evaluate retrieval: did the retrieved context contain useful information for answering the question?

````python
# Define system message for rating context's supportiveness to answer
groundedness_rater_system_message = """\
You are evaluating whether an AI answer is grounded in the provided context.

Score the answer from 1 to 5:
1 = Mostly unsupported or contradicted by the context
2 = Several important claims are unsupported
3 = Partially supported, with some unsupported claims
4 = Almost entirely supported, with only minor issues
5 = Fully supported by the context

Judge only whether the answer is supported by the context. Do not use outside knowledge.

Return exactly:
Score: <1-5>
Reason: <one concise sentence>
"""
````

````python
# Define system message for rating context's relevance to question
relevance_rater_system_message = """\
You are evaluating the quality of retrieved context for answering a question.

Score the context from 1 to 5:
1 = Irrelevant to the question
2 = Mostly irrelevant or missing important information
3 = Partially relevant but incomplete
4 = Mostly relevant and sufficient
5 = Directly relevant and sufficient to answer the question

Evaluate only the relationship between the question and context. Do not use outside knowledge.

Return exactly:
Score: <1-5>
Reason: <one concise sentence>
"""
````

````python
# Generate a RAG response and evaluate its retrieval relevance and groundedness
def evaluate_rag_response(
    user_input,
    k=3,
    temperature=0.0,
    answer_max_tokens=None,
    judge_max_tokens=64
):
    if answer_max_tokens is None:
        answer_max_tokens = default_max_tokens

    # Retrieve the most relevant document chunks
    query_retriever = vectorstore.as_retriever(
        search_kwargs={"k": k}
    )

    relevant_document_chunks = query_retriever.invoke(user_input)

    # Preserve boundaries between retrieved chunks
    context_for_query = "\n\n---\n\n".join(
        document.page_content.strip()
        for document in relevant_document_chunks
    )

    # Build the RAG question
    qna_user_message = qna_user_message_template.format(
        context=context_for_query,
        question=user_input
    )

    # Generate the RAG answer
    answer_output = llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": qna_system_message
            },
            {
                "role": "user",
                "content": qna_user_message
            }
        ],
        max_tokens=answer_max_tokens,
        temperature=temperature
    )

    answer = (
        answer_output["choices"][0]["message"]["content"].strip()
    )

    # Build the common evaluation input
    evaluation_message = user_message_template.format(
        question=user_input,
        context=context_for_query,
        answer=answer
    )

    # Evaluate generation groundedness
    groundedness_output = llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": groundedness_rater_system_message
            },
            {
                "role": "user",
                "content": evaluation_message
            }
        ],
        max_tokens=judge_max_tokens,
        temperature=0.0
    )

    groundedness = (
        groundedness_output["choices"][0]["message"]["content"].strip()
    )

    # Evaluate retrieval relevance
    relevance_output = llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": relevance_rater_system_message
            },
            {
                "role": "user",
                "content": evaluation_message
            }
        ],
        max_tokens=judge_max_tokens,
        temperature=0.0
    )

    relevance = (
        relevance_output["choices"][0]["message"]["content"].strip()
    )

    return {
        "answer": answer,
        "groundedness": groundedness,
        "retrieval_relevance": relevance
    }
````

    - It's adequate for demonstrating the technique, but a 1.7B model is a pretty weak judge of nuanced factual grounding.

    - Numerical score calibration can be inconsistent.
    - The model might misunderstand the evidence while generating and then make the same misunderstanding while judging.
    - A confident score of 5 does not really make an answer objectively grounded or relevant.
    - **An LLM judge cannot reliably detect an error that exceeds the judge's own reasoning ability.**
    - Really in all cases where the generator and evaluator are the same small model, scores should be treated as indicative rather than objective.


## Actionable Insights and Business Recommendations

- **Use current AI libraries and models.**
    - AI tooling changes fast, and old notebooks can create more troubleshooting than learning.
    - Courses should keep examples reasonably current.

- **Cache expensive resources.**
    - Save vector databases and downloaded models so you do not rebuild or redownload them every session.

- **Use enough compute for the job.**
    - Small quantized models can run on CPU, but slowly.
    - GPU makes experimentation dramatically faster.
    - Free Colab can force compromises that would not make sense in production.

- **Choose models for the actual task.**
    - Consider domain, model size, speed, memory, and quantization.
    - A fast small model can be useful for development, while a stronger model can be reserved for final testing.

- **Take embedding models seriously.**
    - Retrieval quality determines what evidence the LLM sees.
    - Domain-specific embeddings can improve specialized RAG.
    - Make sure the embedding model supports your chunk length.

- **Know what the main parameters do.**
    - `n_ctx`: context size and memory use.
    - `n_batch`: prompt-processing speed.
    - `n_gpu_layers`: GPU offloading.
    - `n_threads`: CPU performance.
    - `max_tokens`: output length and runtime.
    - `temperature`, `top_p`, `top_k`: sampling behavior.
    - Retrieval `k`, `chunk_size`, and `chunk_overlap`: how much evidence is retrieved and how it is divided.

- **More context is not always better.**
    - A few strong chunks often worked better than many weaker ones.

- **Clean documents before embedding them.**
    - Remove headers, footers, page numbers, emails, and other PDF noise.
    - Boilerplate wastes context and can leak into responses.

- **RAG helps, but it does not eliminate mistakes.**
    - Good retrieval produced much better answers.
    - Bad retrieval sometimes produced confident but irrelevant ones.
    - RAG can also faithfully repeat outdated information from its source.

- **Keep prompts simple and purposeful.**
    - Clear roles and constraints helped.
    - Repeating the same idea under extra sections usually did not.
    - For factual QA, concise prompts and low temperature worked best.

- **Parameter tuning is not fine-tuning.**
    - Changing prompts, `temperature`, `k`, or `max_tokens` is inference-time tuning.
    - Real fine-tuning changes model or adapter weights through training.

- **Treat LLM-as-a-judge as a rough signal.**
    - Groundedness and retrieval relevance are useful checks, but a small model judging itself is not objective.

- **Run experiments in a controlled way.**
    - Keep questions fixed and change one meaningful variable at a time.
    - Compare accuracy, grounding, retrieval quality, and runtime, not just how polished the answer sounds.
