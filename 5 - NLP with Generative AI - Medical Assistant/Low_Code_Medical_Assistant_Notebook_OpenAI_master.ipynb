### System and User Prompt Template

Prompts guide the model to generate accurate responses. Here, we define two parts:

    1. The system message describing the assistant's role.
    2. A user message template including context and the question.

````python
# Define the system prompt for the model
qna_system_message = """_____""" #Complete the code to define the system message describing the assistant's role, response guidelines, and fallback behavior
````

````python
# Define the user message template
qna_user_message_template = """_____""" #Complete the code to define the user message template, including the ###Context and ###Question sections
````

### Response Function

````python
def generate_rag_response(user_input, k=3, max_tokens=128, temperature=0, top_p=0.95):
    global qna_system_message, qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.invoke(user_input)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    # Generate the response via the OpenAI client (GL Proxy)
    try:
        response = client.chat.completions.create(
            model=_____,                                          #Complete the code to pass the model to be used
            messages=[
                {"role": "system", "content": _____},              #Complete the code to pass the system message
                {"role": "user", "content": _____},                #Complete the code to pass the user message
            ],
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
        )

        # Extract and return the model's response
        response = response.choices[0].message.content.strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
````

## Output Evaluation

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answers generated to the questions from the previous section.

- We use a **different model as the judge** (`judge_llm`) from the one used for answer generation (`llm`), via the GL Proxy. This avoids the generator model rating its own output, which can otherwise inflate scores due to self-preference bias.

````python
groundedness_rater_system_message = """_____""" #Complete the code to define the groundedness evaluation prompt, including the rating scale (1-5) and output format
````

````python
relevance_rater_system_message = """_____""" #Complete the code to define the relevance evaluation prompt, including the rating scale (1-5) and output format
````

````python
user_message_template = """
###Question
{question}

###Context
{context}

###Answer
{answer}
"""
````

````python
def generate_ground_relevance_response(user_input, k=3, max_tokens=128, temperature=0, top_p=0.95):
    global qna_system_message, qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.invoke(user_input)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.format(context=context_for_query, question=user_input)

    # Generate the answer via the OpenAI client (GL Proxy) using the GENERATION model (llm)
    answer = client.chat.completions.create(
        model=_____,                                          #Complete the code to pass the generation model (llm)
        messages=[
            {"role": "system", "content": qna_system_message},
            {"role": "user", "content": user_message},
        ],
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
    ).choices[0].message.content

    filled_prompt = user_message_template.format(context=context_for_query, question=user_input, answer=answer)

    # Groundedness score -- rated using a separate JUDGE model (judge_llm), not the generation model
    response_1 = client.chat.completions.create(
        model=_____,                                          #Complete the code to pass the judge model (judge_llm)
        messages=[
            {"role": "system", "content": _____},              #Complete the code to pass the groundedness rater system message
            {"role": "user", "content": filled_prompt},
        ],
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
    )

    # Relevance score -- rated using a separate JUDGE model (judge_llm), not the generation model
    response_2 = client.chat.completions.create(
        model=_____,                                          #Complete the code to pass the judge model (judge_llm)
        messages=[
            {"role": "system", "content": _____},              #Complete the code to pass the relevance rater system message
            {"role": "user", "content": filled_prompt},
        ],
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
    )

    return response_1.choices[0].message.content, response_2.choices[0].message.content
````
